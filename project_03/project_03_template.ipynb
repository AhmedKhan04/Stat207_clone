## 1. Introduction

```python
#Imports here
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import statsmodels.formula.api as smf
import statsmodels.api as sm
from sklearn.metrics import roc_auc_score
```

```python
df = pd.read_csv('video_games .csv',encoding = 'unicode-escape')
pd.set_option('display.max_columns', None)
df.head()
```

## 2. Linear Regression Analytical Tasks

What is the relationship between review_score and US Sales after controlling for Year released, Review Score, and Used price, and sequel both in the sample and in the underlying population? How well does this model predict US sales for new video games?

```python
model_df = df[
    ['US Sales (millions)', 'YearReleased', 'Review Score', 'Usedprice', 'Sequel']
].dropna()

train_df, test_df = train_test_split(
    model_df,
    test_size=0.2,
    random_state=207
)
```

```python
model = smf.ols(
    'Q("US Sales (millions)") ~ YearReleased + Q("Review Score") + Usedprice + Sequel',
    data=train_df
).fit()
print(model.summary())
```

US Sales_hat = 190.2154
             - 0.0955(YearReleased)
             + 0.0190(Review Score)
             + 0.0276(UsedPrice)
             + 0.1345(Sequel)


The slope coefficient for Review Score is 0.0190. This means that, holding YearReleased, UsedPrice, and Sequel constant, a one-point increase in review score is associated with an estimated increase of 0.0190 million US sales, or about 19,000 additional US sales, on average.

The 95% confidence interval for the Review Score slope is:
(0.015, 0.023)


```python
train_df = train_df.copy()
train_df['fitted'] = model.fittedvalues
train_df['residuals'] = model.resid
```

```python
sns.scatterplot(x='fitted', y='residuals', data=train_df)
plt.axhline(y=0, color='red')
plt.xlabel('Fitted Values')
plt.ylabel('Residuals')
plt.title('Residuals vs Fitted Values')
plt.show()
```

```python
sm.qqplot(train_df['residuals'], line='45')
plt.title('Q-Q Plot of Residuals')
plt.show()
```

The percent of variability of US Sales that can be explained by this model is 15.3%. This is  low because most of the variation in US sales is still unexplained by the model. This means YearReleased, Review Score, UsedPrice, and Sequel have some relationship with US sales, but they are not enough to strongly predict US sales by themselves.

```python
test_pred = model.predict(test_df)

rmse = np.sqrt(((test_df['US Sales (millions)'] - test_pred) ** 2).mean())

rmse
```

The test RMSE is 1.328 million. Since US Sales (millions) is measured in millions, this means the model’s predicted US sales are typically off by about 1.328 million copies sold for games in the test data. No this is not low as 1.328 million copies can become upwards of 10 millions of dollars off when factoring in Sale price.

## 3. Logistic Regression Analytical Tasks

How do YearReleased, Review Score, UsedPrice, and Sequel relate to the log-odds of a video game having high US sales in the sample training data? How does a classifier built from this logistic regression model perform on new video games?

```python
median_sales = df['US Sales (millions)'].median()

df['High_US_Sales'] = (df['US Sales (millions)'] > median_sales).astype(int)

df['High_US_Sales'].value_counts()
```

```python
log_model_df = df[
    ['High_US_Sales', 'YearReleased', 'Review Score', 'Usedprice', 'Sequel']
].dropna()

train_log_df, test_log_df = train_test_split(
    log_model_df,
    test_size=0.2,
    random_state=207
)
```

```python
log_model = smf.logit(
    'High_US_Sales ~ YearReleased + Q("Review Score") + Usedprice + Sequel',
    data=train_log_df
).fit()

log_model.summary()
```

```python
log_model.prsquared
```

YearReleased has a coefficient of -0.0114, so newer release years are associated with a lower probability of high US sales, holding the other variables constant. However, its p-value is 0.822, so this relationship does not appear statistically significant.

Review Score has a coefficient of 0.0695, so higher review scores are associated with a higher probability of high US sales, holding the other variables constant. Its p-value is approximately 0.000, so this appears to be a statistically significant positive relationship.

Usedprice has a coefficient of 0.0007, so higher used prices are associated with a slightly higher probability of high US sales, holding the other variables constant. Its p-value is 0.939, so this relationship does not appear statistically significant.

Sequel has a coefficient of 0.4560, so sequels are associated with a higher probability of high US sales than non-sequels, holding the other variables constant. Its p-value is approximately 0.000, so this appears to be statistically significant.

```python
train_probs = log_model.predict(train_log_df)

auc = roc_auc_score(train_log_df['High_US_Sales'], train_probs)
auc
```

The AUC is 0.76585. AUC measures how well the model separates high-US-sales games from non-high-US-sales games. A value near 0.5 means the model is close to random guessing, while a value closer to 1 means stronger classification ability. Since this AUC is about 0.766, the model has moderately strong classification ability. It is much better than random guessing, but it is not close enough to 1 to be considered extremely strong.

```python
test_probs = log_model.predict(test_log_df)

test_log_df = test_log_df.copy()
test_log_df['predicted'] = (test_probs >= 0.5).astype(int)
```

```python
TP = ((test_log_df['High_US_Sales'] == 1) & (test_log_df['predicted'] == 1)).sum()
TN = ((test_log_df['High_US_Sales'] == 0) & (test_log_df['predicted'] == 0)).sum()
FP = ((test_log_df['High_US_Sales'] == 0) & (test_log_df['predicted'] == 1)).sum()
FN = ((test_log_df['High_US_Sales'] == 1) & (test_log_df['predicted'] == 0)).sum()

accuracy = (TP + TN) / (TP + TN + FP + FN)
sensitivity = TP / (TP + FN)
specificity = TN / (TN + FP)

print(F"Accuracy:{accuracy}, Sensitivity: {sensitivity}, Specificity: {specificity}")
```

Accuracy is the proportion of all test games classified correctly.

Sensitivity is the proportion of actual high-US-sales games that were correctly classified as high sales.

Specificity is the proportion of actual non-high-US-sales games that were correctly classified as not high sales.

In this context, I would want to minimize false positives. A false positive means the model predicts that a game will have high US sales, but the game actually does not have high US sales. This could be costly for a company because it might over-invest in advertising, production, distribution, or inventory for a game that does not end up selling well.

I am somewhat satisfied with this classifier, but not completely. The accuracy of about 70.06% and AUC of 0.76585 suggest that the model performs better than random guessing and has moderate classification ability. However, since I care about minimizing false positives, the specificity of 66.10% is important. This means the model correctly identifies about two-thirds of the non-high-sales games, but it still incorrectly labels some lower-sales games as high-sales games. Because of that, the classifier is useful but not strong enough to rely on by itself for major business decisions.
